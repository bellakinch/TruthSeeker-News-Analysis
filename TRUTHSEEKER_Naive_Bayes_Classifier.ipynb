```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from google.colab import drive
import pandas as pd
import numpy as np

#  ----- Load and clean data ------
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/Features_For_Traditional_ML_Techniques (1).csv')
print(df.head())
print(df.shape)
print(df.describe(include="all"))

# Remove any rows with missing values
df = df.dropna(subset=["BinaryNumTarget"]).copy()

# Create Y target variable / y = whether the statement is TRUE (1) or FALSE (0)
y = df["BinaryNumTarget"]

# Show how many true and false statements
print("\nTarget Distribution:")
print(y.value_counts())

# ------- MODEL 1 : What they said -------
what_features = [
    "DATE_percentage",
    "total_count",
    "PERSON_percentage",
    "LANGUAGE_percentage",
    "PERCENT_percentage",
    "LAW_percentage",
    "PRODUCT_percentage",
    "EVENT_percentage",
    "ORG_percentage",
    "GPE_percentage"
]

# Create the feature dataset
X_what = df[what_features].apply(
    pd.to_numeric,
    errors="coerce"
)

# Fill missing values
X_what = X_what.fillna(0)

# Convert each feature so the minimum value is 0
X_what = X_what - X_what.min()

# Get rid of any negative values
X_what = X_what.clip(lower=0)

# Apply log transformation no 0s
X_what = np.log1p(X_what)

# Split the data training/ testing
X_train_what, X_test_what, y_train_what, y_test_what = train_test_split(
    X_what,
    y,
    test_size=0.30,
    random_state=42
)

# Create Multinomial Naive Bayes pipeline
model_what = make_pipeline(
    MultinomialNB()
)

# Teaches Naive Bayes model using training data
model_what.fit(X_train_what, y_train_what)

# Make predictions
y_pred_what = model_what.predict(X_test_what)

# Evaluate the model
# Evaluate the model
print("\nMODEL 1: What the person says")

print("\nAccuracy:")
print(accuracy_score(y_test_what, y_pred_what))

print("\nClassification Report:")
print(classification_report(y_test_what, y_pred_what))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_what, y_pred_what))

# ------- MODEL 2 : How they say it ----------
how_features = [
    "Word count",
    "Max word length",
    "Min word length",
    "Average word length",
    "present_verbs",
    "past_verbs",
    "exclamation",
    "questions",
    "long_word_freq",
    "short_word_freq"
]

# Create the feature dataset
X_how = df[how_features].apply(
    pd.to_numeric,
    errors="coerce"
)

# Fill missing values
X_how = X_how.fillna(0)

# Shift each feature so the minimum value is 0
X_how = X_how - X_how.min()

# Apply log so no negative values
X_how = np.log1p(X_how)

# Make absolutely sure there are no negative values
X_how = X_how.clip(lower=0)

# Check the minimum value
print("\nMinimum value after log transformation:")
print(X_how.min().min())

# Split the data into training and testing sets
X_train_how, X_test_how, y_train_how, y_test_how = train_test_split(
    X_how,
    y,
    test_size=0.30,
    random_state=42
)

# Create Multinomial Naive Bayes pipeline
model_how = make_pipeline(
    MultinomialNB()
)

# Train the model
model_how.fit(X_train_how, y_train_how)

# Make predictions
y_pred_how = model_how.predict(X_test_how)

# Evaluate the model
print("\nMODEL 2: How the person says it")

print("\nAccuracy:")
print(accuracy_score(y_test_how, y_pred_how))

print("\nClassification Report:")
print(classification_report(y_test_how, y_pred_how))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_how, y_pred_how))
```
